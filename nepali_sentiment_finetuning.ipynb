## SECTION 2: Baseline Evaluation (BEFORE Fine-Tuning)

**Purpose**: Establish zero-shot performance of the pre-trained XLM-RoBERTa model on Nepali sentiment.

**Why this matters**: This is our baseline. Without this, we can't prove that fine-tuning actually improved anything. These numbers will be compared directly against post-training results to demonstrate measurable impact.

```python
# Load pre-trained XLM-RoBERTa model (zero-shot, no fine-tuning yet)
# XLM-RoBERTa is chosen because it's multilingual and handles Devanagari script well

MODEL_NAME = "xlm-roberta-base"

print(f"Loading {MODEL_NAME} for baseline evaluation...")
baseline_tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
baseline_model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=2,
    ignore_mismatched_sizes=True  # The pre-trained model doesn't have a classification head
)
baseline_model.to(device)
baseline_model.eval()

print("âœ“ Model loaded successfully")
```

```python
# Sample 200 random test examples for baseline evaluation
# Using a fixed sample so we can compare baseline vs fine-tuned on identical data

np.random.seed(42)  # For reproducibility
test_sample_indices = np.random.choice(len(test_df), size=200, replace=False)
test_sample_df = test_df.iloc[test_sample_indices].reset_index(drop=True)

print(f"Selected {len(test_sample_df)} test examples for evaluation")
print(f"Distribution: {test_sample_df['label'].value_counts().sort_index().to_dict()}")
```

```python
# Run baseline inference
# Using batch processing to speed up evaluation

def predict_batch(model, tokenizer, texts, batch_size=32):
    """
    Run inference on a batch of texts.
    Returns predictions as a list of label indices (0 or 1).
    """
    model.eval()
    all_predictions = []

    with torch.no_grad():
        for i in range(0, len(texts), batch_size):
            batch_texts = texts[i:i+batch_size]

            # Tokenize
            inputs = tokenizer(
                batch_texts,
                padding=True,
                truncation=True,
                max_length=128,
                return_tensors='pt'
            ).to(device)

            # Forward pass
            outputs = model(**inputs)
            predictions = torch.argmax(outputs.logits, dim=-1)
            all_predictions.extend(predictions.cpu().numpy())

    return all_predictions

print("Running baseline inference on 200 test samples...")
baseline_predictions = predict_batch(
    baseline_model,
    baseline_tokenizer,
    test_sample_df['text'].tolist(),
    batch_size=32
)
print("âœ“ Baseline inference complete")
```

```python
# Calculate baseline metrics
true_labels = test_sample_df['label'].tolist()

baseline_accuracy = accuracy_score(true_labels, baseline_predictions)
baseline_f1 = f1_score(true_labels, baseline_predictions, average='binary')
baseline_cm = confusion_matrix(true_labels, baseline_predictions)

print("=" * 60)
print("ðŸ“Š BASELINE PERFORMANCE (BEFORE FINE-TUNING)")
print("=" * 60)
print(f"\nAccuracy: {baseline_accuracy*100:.2f}%")
print(f"F1 Score: {baseline_f1:.4f}")
print(f"\nConfusion Matrix:")
print(baseline_cm)
print(f"\nClassification Report:")
print(classification_report(true_labels, baseline_predictions, target_names=['Negative', 'Positive']))

# Visualize confusion matrix
plt.figure(figsize=(8, 6))
sns.heatmap(baseline_cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Negative', 'Positive'],
            yticklabels=['Negative', 'Positive'])
plt.title('Baseline Confusion Matrix (Zero-Shot)')
plt.ylabel('True Label')
plt.xlabel('Predicted Label')
plt.show()

# Store baseline results for later comparison
baseline_results = {
    'accuracy': baseline_accuracy,
    'f1': baseline_f1,
    'predictions': baseline_predictions
}

print("\nâœ“ Baseline evaluation complete. These numbers will be compared against fine-tuned model.")
```

## SECTION 3: Fine-Tuning with Manual Training Loop

**Purpose**: Fine-tune XLM-RoBERTa on the Nepali sentiment dataset using a manual training loop.

**Why manual loop instead of Trainer API**: This demonstrates understanding of the actual training mechanics (forward pass, loss calculation, backpropagation, optimization) rather than just calling .train().

```python
# Tokenize the dataset
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

def tokenize_function(examples):
    return tokenizer(
        examples['sentences'],  # Use 'sentences' - the original column name
        padding='max_length',
        truncation=True,
        max_length=128
    )

print("Tokenizing datasets...")
tokenized_train = dataset['train'].map(tokenize_function, batched=True, remove_columns=['sentences'])
tokenized_test = dataset['test'].map(tokenize_function, batched=True, remove_columns=['sentences'])

# Set format for PyTorch
tokenized_train.set_format('torch', columns=['input_ids', 'attention_mask', 'sentiment'])
tokenized_test.set_format('torch', columns=['input_ids', 'attention_mask', 'sentiment'])

# Rename 'sentiment' to 'label' for consistency with training code
tokenized_train = tokenized_train.rename_column('sentiment', 'label')
tokenized_test = tokenized_test.rename_column('sentiment', 'label')

print("âœ“ Tokenization complete")
```

```python
# Loss function setup (FIXED: removed class weights - dataset is balanced)
from torch import nn

# Dataset is balanced (49.25% / 50.75%), so no class weighting needed
criterion = nn.CrossEntropyLoss()
print("âœ“ Loss function: CrossEntropyLoss (no class weighting)")
```

```python
# Initialize model for fine-tuning
model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=2,
    ignore_mismatched_sizes=True
)
model.to(device)

# FIX: Remove torchvision dependency
tokenized_train = tokenized_train.with_format("torch", columns=['input_ids', 'attention_mask', 'label'], output_all_columns=False)
tokenized_test = tokenized_test.with_format("torch", columns=['input_ids', 'attention_mask', 'label'], output_all_columns=False)

# Create DataLoaders
from torch.utils.data import DataLoader

train_dataloader = DataLoader(tokenized_train, batch_size=16, shuffle=True)
test_dataloader = DataLoader(tokenized_test, batch_size=32, shuffle=False)
```

```python
# Set up optimizer and learning rate scheduler
from torch.optim import AdamW
from transformers import get_linear_schedule_with_warmup

optimizer = AdamW(model.parameters(), lr=2e-5)

# Calculate total training steps
NUM_EPOCHS = 2  # Reduced from 3 - converges faster with fixes
total_steps = len(train_dataloader) * NUM_EPOCHS
warmup_steps = int(0.1 * total_steps)  # 10% warmup

# Add linear warmup scheduler for stable training
scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=warmup_steps,
    num_training_steps=total_steps
)

print(f"\nTraining configuration:")
print(f"  Epochs: {NUM_EPOCHS}")
print(f"  Batch size: 16")
print(f"  Learning rate: 2e-5")
print(f"  Optimizer: AdamW")
print(f"  Scheduler: Linear warmup ({warmup_steps} steps)")
print(f"  Gradient accumulation: 4 steps")
print(f"  Gradient clipping: 1.0")
print(f"  Total training steps: {total_steps}")
```

```python
# Clear torchvision from memory to avoid VideoReader import error
import sys
if 'torchvision' in sys.modules:
    del sys.modules['torchvision']

# Recreate DataLoaders without torchvision interference
from torch.utils.data import DataLoader

# Convert to list of dicts to avoid format issues
train_data = [{
    'input_ids': tokenized_train[i]['input_ids'],
    'attention_mask': tokenized_train[i]['attention_mask'],
    'label': tokenized_train[i]['label']
} for i in range(len(tokenized_train))]

test_data = [{
    'input_ids': tokenized_test[i]['input_ids'],
    'attention_mask': tokenized_test[i]['attention_mask'],
    'label': tokenized_test[i]['label']
} for i in range(len(tokenized_test))]

# Create new DataLoaders from the converted data
train_dataloader = DataLoader(train_data, batch_size=16, shuffle=True)
test_dataloader = DataLoader(test_data, batch_size=32, shuffle=False)

print("âœ“ DataLoaders recreated without torchvision")
print(f"Training batches: {len(train_dataloader)}")
print(f"Test batches: {len(test_dataloader)}")
```

```python
# MANUAL TRAINING LOOP (FIXED)
# Added: gradient accumulation, clipping, warmup scheduler, diagnostics

print("\n" + "=" * 60)
print("STARTING FINE-TUNING")
print("=" * 60)

training_losses = []
ACCUMULATION_STEPS = 4

for epoch in range(NUM_EPOCHS):
    model.train()
    epoch_loss = 0
    accumulated_loss = 0
    progress_bar = tqdm(train_dataloader, desc=f"Epoch {epoch+1}/{NUM_EPOCHS}")

    optimizer.zero_grad()  # Initialize gradients

    for step, batch in enumerate(progress_bar):
        # Move batch to device
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        labels = batch['label'].to(device)

        # Forward pass
        outputs = model(input_ids=input_ids, attention_mask=attention_mask)
        logits = outputs.logits

        # Calculate loss
        loss = criterion(logits, labels)

        # Scale loss for gradient accumulation
        loss = loss / ACCUMULATION_STEPS

        # Backward pass
        loss.backward()

        # Accumulate loss for reporting (unscaled)
        accumulated_loss += loss.item() * ACCUMULATION_STEPS
        epoch_loss += loss.item() * ACCUMULATION_STEPS

        # Update weights every ACCUMULATION_STEPS
        if (step + 1) % ACCUMULATION_STEPS == 0:
            # Gradient clipping
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)

            # Optimizer step
            optimizer.step()
            scheduler.step()
            optimizer.zero_grad()

            # Update progress bar
            progress_bar.set_postfix({
                'loss': f'{accumulated_loss / ACCUMULATION_STEPS:.4f}',
                'lr': f'{scheduler.get_last_lr()[0]:.2e}'
            })
            accumulated_loss = 0

        # Diagnostic prints for first epoch
        if epoch == 0 and step in [0, 9, 49]:
            current_loss = loss.item() * ACCUMULATION_STEPS
            print(f"\n  Batch {step+1}: loss = {current_loss:.4f}")

    # Handle last batch if not perfectly divisible
    if (step + 1) % ACCUMULATION_STEPS != 0:
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()
        scheduler.step()
        optimizer.zero_grad()

    # Calculate average loss for epoch
    avg_loss = epoch_loss / len(train_dataloader)
    training_losses.append(avg_loss)

    print(f"\nâœ“ Epoch {epoch+1} complete - Average Loss: {avg_loss:.4f}")

    # Clear GPU cache
    torch.cuda.empty_cache()

print("\n" + "=" * 60)
print("FINE-TUNING COMPLETE")
print("=" * 60)
```

```python
# Plot training loss curve
plt.figure(figsize=(10, 6))
plt.plot(range(1, NUM_EPOCHS+1), training_losses, marker='o', linewidth=2, markersize=8)
plt.title('Training Loss Over Epochs', fontsize=14, fontweight='bold')
plt.xlabel('Epoch')
plt.ylabel('Average Loss')
plt.grid(True, alpha=0.3)
plt.xticks(range(1, NUM_EPOCHS+1))
plt.tight_layout()
plt.show()

print("\nTraining losses by epoch:")
for i, loss in enumerate(training_losses, 1):
    print(f"  Epoch {i}: {loss:.4f}")
```

## SECTION 5: Error Analysis

**Purpose**: Honest evaluation of model performance - what improved, and what still fails.

**Why this matters**: This shows intellectual honesty and critical thinking. Anyone can report accuracy numbers, but understanding *where* and *why* a model fails demonstrates deep ML maturity.

## SECTION 6: Simple Inference Function

**Purpose**: Create a simple, reusable function for predicting sentiment on new Nepali text.



```python
def predict_sentiment(text, return_confidence=True):
    """
    Predict sentiment for a single Nepali text string.

    Args:
        text (str): Nepali text to analyze
        return_confidence (bool): Whether to return confidence score

    Returns:
        dict: Contains 'sentiment' (positive/negative) and optional 'confidence'
    """
    # Tokenize
    inputs = tokenizer(
        text,
        padding=True,
        truncation=True,
        max_length=128,
        return_tensors='pt'
    ).to(device)

    # Get prediction
    model.eval()
    with torch.no_grad():
        outputs = model(**inputs)
        logits = outputs.logits
        probabilities = torch.softmax(logits, dim=-1)
        prediction = torch.argmax(logits, dim=-1).item()
        confidence = probabilities[0][prediction].item()

    sentiment = "positive" if prediction == 1 else "negative"

    if return_confidence:
        return {
            'sentiment': sentiment,
            'confidence': confidence,
            'label': prediction
        }
    else:
        return {'sentiment': sentiment}

print("âœ“ predict_sentiment() function defined")
```
